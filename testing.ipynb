```python
import numpy as np
from alfworld.agents.environment.alfred_tw_env import AlfredTWEnv
```

```python
config = {
    'env': {
        'type': 'AlfredTWEnv',
        'alfred': {
            'data_path': '~/.cache/alfworld/json_2.1.1'
        }
    },
    'general': {
        'random_seed': 42
    }
}


env = AlfredTWEnv(config, train_eval='train')
env = env.init_env(batch_size=1)

# interact
obs, info = env.reset()


for _ in range(10):
    admissible_commands = list(info['admissible_commands'])
    random_actions = [np.random.choice(admissible_commands[0])]
    
    obs, scores, dones, infos = env.step(random_actions)
    print(f"Action: {random_actions[0]}")
    print(f"Obs: {obs[0][:200]}...")
    print(f"Score: {scores[0]}, Done: {dones[0]}\n")
    
    if dones[0]:
        print("Episode finished!")
        break
    
    info = infos
```
